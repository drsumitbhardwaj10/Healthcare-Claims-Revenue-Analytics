```python
# --------------------------------------------------
# Record count validation
# --------------------------------------------------

print("Record Count Validation")
print("=" * 70)

record_counts = {}

for name, file_path in files.items():

    # Count rows without loading the complete dataset into memory
    row_count = sum(1 for _ in open(file_path, encoding="utf-8")) - 1

    record_counts[name] = row_count

    print(f"{name:20} | {row_count:,} records")
```

```python
# --------------------------------------------------
# Duplicate and identifier validation
# --------------------------------------------------

print("Duplicate and Identifier Validation")
print("=" * 70)

# --------------------------------------------------
# Beneficiary datasets
# --------------------------------------------------

beneficiary_files = {
    "beneficiary_2008": files["beneficiary_2008"],
    "beneficiary_2009": files["beneficiary_2009"],
    "beneficiary_2010": files["beneficiary_2010"],
}

for name, file_path in beneficiary_files.items():

    df = pd.read_csv(file_path, usecols=["DESYNPUF_ID"])

    total_rows = len(df)
    unique_ids = df["DESYNPUF_ID"].nunique()
    duplicate_rows = total_rows - unique_ids

    print(f"\n{name}")
    print(f"Total records       : {total_rows:,}")
    print(f"Unique beneficiary IDs: {unique_ids:,}")
    print(f"Duplicate ID rows   : {duplicate_rows:,}")


# --------------------------------------------------
# Claims datasets
# --------------------------------------------------

claim_files = {
    "inpatient": files["inpatient"],
    "outpatient": files["outpatient"],
}

for name, file_path in claim_files.items():

    unique_ids = set()
    duplicate_ids = set()
    total_rows = 0

    for chunk in pd.read_csv(
        file_path,
        usecols=["CLM_ID"],
        chunksize=100_000
    ):

        for claim_id in chunk["CLM_ID"].dropna():

            total_rows += 1

            if claim_id in unique_ids:
                duplicate_ids.add(claim_id)
            else:
                unique_ids.add(claim_id)

    print(f"\n{name}")
    print(f"Total claim records : {total_rows:,}")
    print(f"Unique claim IDs    : {len(unique_ids):,}")
    print(f"Duplicate claim IDs : {len(duplicate_ids):,}")
```

### Claims Files

Repeated `CLM_ID` values were identified:

- Inpatient: 68 duplicate claim-ID occurrences
- Outpatient: 10,975 duplicate claim-ID occurrences

These repeated claim IDs are not automatically classified as erroneous duplicates. The claims datasets contain a `SEGMENT` field, so the relationship between `CLM_ID` and `SEGMENT` must be investigated before any records are removed.

### Data Quality Decision

No records will be deleted at this stage.

The repeated claim IDs will be investigated during the detailed data-quality assessment to determine whether they represent:

- multiple segments belonging to the same claim,
- legitimate repeated claim records, or
- actual duplicate records.

```python
# --------------------------------------------------
# Investigate CLM_ID + SEGMENT uniqueness
# --------------------------------------------------

print("CLM_ID + SEGMENT Validation")
print("=" * 70)

for name, file_path in claim_files.items():

    pair_counts = {}

    for chunk in pd.read_csv(
        file_path,
        usecols=["CLM_ID", "SEGMENT"],
        chunksize=100_000
    ):
        chunk = chunk.dropna(subset=["CLM_ID"])

        for claim_id, segment in zip(
            chunk["CLM_ID"],
            chunk["SEGMENT"]
        ):
            key = (claim_id, segment)
            pair_counts[key] = pair_counts.get(key, 0) + 1

    total_records = sum(pair_counts.values())
    unique_pairs = len(pair_counts)

    duplicate_pairs = sum(
        count > 1 for count in pair_counts.values()
    )

    duplicate_rows = sum(
        count - 1
        for count in pair_counts.values()
        if count > 1
    )

    print(f"\n{name}")
    print(f"Total claim records       : {total_records:,}")
    print(f"Unique CLM_ID + SEGMENT  : {unique_pairs:,}")
    print(f"Duplicate pairs           : {duplicate_pairs:,}")
    print(f"Duplicate rows            : {duplicate_rows:,}")
```

  - 0 duplicate combinations

- Outpatient Claims:

### Interpretation

### Data Quality Decision

The composite key `CLM_ID + SEGMENT` should be considered when performing claim-level record validation and analysis.

No records will be removed based on the repeated `CLM_ID` values.

## Missing Value Validation

The next validation step assesses the completeness of the datasets by identifying missing values in each field.

Missing values are not automatically treated as errors. In healthcare claims data, some fields may legitimately be blank depending on the claim type, beneficiary status, or whether a particular field applies to a record.

The purpose of this step is to identify fields requiring further investigation during the Data Quality Assessment phase.

> **Important:** No missing values will be filled, removed, or otherwise modified during this validation step.

```python
# --------------------------------------------------
# Missing Value Validation
# --------------------------------------------------

print("Missing Value Validation")
print("=" * 70)

for name, file_path in files.items():

    print(f"\n{name}")
    print("-" * 70)

    missing_counts = {}

    for chunk in pd.read_csv(
        file_path,
        chunksize=100_000
    ):
        chunk_missing = chunk.isna().sum()

        for column, count in chunk_missing.items():
            missing_counts[column] = (
                missing_counts.get(column, 0) + count
            )

    total_records = sum(1 for _ in open(file_path, encoding="utf-8")) - 1

    missing_df = pd.DataFrame({
        "Missing_Count": missing_counts
    })

    missing_df["Missing_Percentage"] = (
        missing_df["Missing_Count"] / total_records * 100
    )

    missing_df = (
        missing_df[missing_df["Missing_Count"] > 0]
        .sort_values("Missing_Count", ascending=False)
    )

    if missing_df.empty:
        print("No missing values identified.")
    else:
        display(missing_df)
```

Because a death date is only expected for beneficiaries with a recorded death event, this high level of missingness is not automatically considered a data-quality error.

This issue will be investigated during the data-quality assessment and cleaning stages.

### Data Quality Decision

Missing values will not be removed or imputed during the validation stage.

Fields with very high or complete missingness will be evaluated for removal during the cleaning process based on their analytical relevance and business requirements.

## Date Validity and Chronological Validation

This validation examines the date fields in the beneficiary and claims datasets.

The assessment checks:

- Whether beneficiary birth dates can be interpreted as valid dates
- Whether beneficiary death dates can be interpreted as valid dates
- Whether claim start and end dates can be interpreted as valid dates
- Whether `CLM_FROM_DT` occurs on or before `CLM_THRU_DT`

Invalid or logically inconsistent dates will be documented for investigation during the cleaning stage.

> **Important:** No date values will be modified during this validation step.

```python
# --------------------------------------------------
# Date Validity and Chronological Validation
# --------------------------------------------------

print("Date Validity and Chronological Validation")
print("=" * 70)


# --------------------------------------------------
# Beneficiary Date Validation
# --------------------------------------------------

print("\nBENEFICIARY DATE VALIDATION")
print("-" * 70)

for name in [
    "beneficiary_2008",
    "beneficiary_2009",
    "beneficiary_2010"
]:

    file_path = files[name]

    df = pd.read_csv(
        file_path,
        usecols=["BENE_BIRTH_DT", "BENE_DEATH_DT"],
        low_memory=False
    )

    birth_dates = pd.to_datetime(
        df["BENE_BIRTH_DT"].astype("Int64").astype("string"),
        format="%Y%m%d",
        errors="coerce"
    )

    death_dates = pd.to_datetime(
        df["BENE_DEATH_DT"].astype("Int64").astype("string"),
        format="%Y%m%d",
        errors="coerce"
    )

    invalid_birth = (
        df["BENE_BIRTH_DT"].notna() &
        birth_dates.isna()
    ).sum()

    invalid_death = (
        df["BENE_DEATH_DT"].notna() &
        death_dates.isna()
    ).sum()

    death_before_birth = (
        death_dates.notna() &
        birth_dates.notna() &
        (death_dates < birth_dates)
    ).sum()

    print(f"\n{name}")
    print(f"Invalid birth dates       : {invalid_birth:,}")
    print(f"Invalid death dates       : {invalid_death:,}")
    print(f"Death before birth        : {death_before_birth:,}")


# --------------------------------------------------
# Claims Date Validation
# --------------------------------------------------

print("\nCLAIMS DATE VALIDATION")
print("-" * 70)

for name in ["inpatient", "outpatient"]:

    file_path = files[name]

    date_columns = [
        "CLM_FROM_DT",
        "CLM_THRU_DT"
    ]

    df = pd.read_csv(
        file_path,
        usecols=date_columns,
        low_memory=False
    )

    from_dates = pd.to_datetime(
        df["CLM_FROM_DT"].astype("Int64").astype("string"),
        format="%Y%m%d",
        errors="coerce"
    )

    thru_dates = pd.to_datetime(
        df["CLM_THRU_DT"].astype("Int64").astype("string"),
        format="%Y%m%d",
        errors="coerce"
    )

    invalid_from = (
        df["CLM_FROM_DT"].notna() &
        from_dates.isna()
    ).sum()

    invalid_thru = (
        df["CLM_THRU_DT"].notna() &
        thru_dates.isna()
    ).sum()

    from_after_thru = (
        from_dates.notna() &
        thru_dates.notna() &
        (from_dates > thru_dates)
    ).sum()

    print(f"\n{name}")
    print(f"Invalid claim start dates : {invalid_from:,}")
    print(f"Invalid claim end dates   : {invalid_thru:,}")
    print(f"Start date after end date : {from_after_thru:,}")
```

## Financial and Numeric Validation

This validation examines key numeric fields used for claims and financial analysis.

The assessment focuses on:

- Claim payment amounts (`CLM_PMT_AMT`)
- Primary payer amounts
- Beneficiary liability amounts
- Inpatient utilization days
- Beneficiary reimbursement amounts

The checks identify:

- Negative values
- Zero values
- Minimum and maximum values
- Basic distribution statistics

Negative values are not automatically considered errors because some financial fields may represent adjustments or offsets. Any unusual values will therefore be investigated before a cleaning decision is made.

> **Important:** No financial values will be modified during this validation step.

```python
# --------------------------------------------------
# Financial and Numeric Validation
# --------------------------------------------------

print("Financial and Numeric Validation")
print("=" * 70)


# --------------------------------------------------
# Claims Financial Fields
# --------------------------------------------------

claim_numeric_columns = [
    "CLM_PMT_AMT",
    "NCH_PRMRY_PYR_CLM_PD_AMT"
]

for name in ["inpatient", "outpatient"]:

    print(f"\n{name}")
    print("-" * 70)

    file_path = files[name]

    summary = {
        column: {
            "count": 0,
            "negative": 0,
            "zero": 0,
            "minimum": None,
            "maximum": None,
            "sum": 0
        }
        for column in claim_numeric_columns
    }

    for chunk in pd.read_csv(
        file_path,
        usecols=claim_numeric_columns,
        chunksize=100_000,
        low_memory=False
    ):

        for column in claim_numeric_columns:

            series = chunk[column].dropna()

            if len(series) == 0:
                continue

            summary[column]["count"] += len(series)
            summary[column]["negative"] += (series < 0).sum()
            summary[column]["zero"] += (series == 0).sum()

            chunk_min = series.min()
            chunk_max = series.max()

            if (
                summary[column]["minimum"] is None
                or chunk_min < summary[column]["minimum"]
            ):
                summary[column]["minimum"] = chunk_min

            if (
                summary[column]["maximum"] is None
                or chunk_max > summary[column]["maximum"]
            ):
                summary[column]["maximum"] = chunk_max

            summary[column]["sum"] += series.sum()

    financial_summary = pd.DataFrame(summary).T

    display(financial_summary)


# --------------------------------------------------
# Inpatient Utilization
# --------------------------------------------------

print("\nINPATIENT UTILIZATION")
print("-" * 70)

inpatient_file = files["inpatient"]

utilization_summary = {
    "count": 0,
    "negative": 0,
    "zero": 0,
    "minimum": None,
    "maximum": None
}

for chunk in pd.read_csv(
    inpatient_file,
    usecols=["CLM_UTLZTN_DAY_CNT"],
    chunksize=100_000,
    low_memory=False
):

    series = chunk["CLM_UTLZTN_DAY_CNT"].dropna()

    if len(series) == 0:
        continue

    utilization_summary["count"] += len(series)
    utilization_summary["negative"] += (series < 0).sum()
    utilization_summary["zero"] += (series == 0).sum()

    chunk_min = series.min()
    chunk_max = series.max()

    if (
        utilization_summary["minimum"] is None
        or chunk_min < utilization_summary["minimum"]
    ):
        utilization_summary["minimum"] = chunk_min

    if (
        utilization_summary["maximum"] is None
        or chunk_max > utilization_summary["maximum"]
    ):
        utilization_summary["maximum"] = chunk_max

display(pd.DataFrame([utilization_summary]))


# --------------------------------------------------
# Beneficiary Reimbursement Fields
# --------------------------------------------------

print("\nBENEFICIARY REIMBURSEMENT FIELDS")
print("-" * 70)

beneficiary_numeric_columns = [
    "MEDREIMB_IP",
    "BENRES_IP",
    "PPPYMT_IP",
    "MEDREIMB_OP",
    "BENRES_OP",
    "PPPYMT_OP",
    "MEDREIMB_CAR",
    "BENRES_CAR",
    "PPPYMT_CAR"
]

beneficiary_file = files["beneficiary_2010"]

beneficiary_summary = {}

for column in beneficiary_numeric_columns:

    values = pd.read_csv(
        beneficiary_file,
        usecols=[column],
        low_memory=False
    )[column].dropna()

    beneficiary_summary[column] = {
        "count": len(values),
        "negative": (values < 0).sum(),
        "zero": (values == 0).sum(),
        "minimum": values.min(),
        "maximum": values.max(),
        "sum": values.sum()
    }

display(pd.DataFrame(beneficiary_summary).T)
```

## Beneficiary-to-Claim Relationship Validation

The beneficiary and claims datasets are related through `DESYNPUF_ID`.

This validation checks whether claim records can be linked to a beneficiary present in the annual beneficiary summary files.

The assessment will identify:

- Claims with a matching beneficiary ID
- Claims whose beneficiary ID is not present in the beneficiary files
- The percentage of claims that can be linked

Because the beneficiary files are provided separately by year, the three annual beneficiary populations will first be combined logically for validation purposes.

> **Important:** No source data will be modified during this validation.

```python
# --------------------------------------------------
# Beneficiary-to-Claim Relationship Validation
# --------------------------------------------------

print("Beneficiary-to-Claim Relationship Validation")
print("=" * 70)


# --------------------------------------------------
# Build beneficiary ID reference set
# --------------------------------------------------

beneficiary_ids = set()

for name in [
    "beneficiary_2008",
    "beneficiary_2009",
    "beneficiary_2010"
]:

    file_path = files[name]

    df = pd.read_csv(
        file_path,
        usecols=["DESYNPUF_ID"],
        low_memory=False
    )

    beneficiary_ids.update(
        df["DESYNPUF_ID"].dropna()
    )

print(f"Unique beneficiary IDs across annual files: {len(beneficiary_ids):,}")


# --------------------------------------------------
# Validate claim relationships
# --------------------------------------------------

for name in ["inpatient", "outpatient"]:

    file_path = files[name]

    total_claims = 0
    matched_claims = 0
    unmatched_claims = 0

    for chunk in pd.read_csv(
        file_path,
        usecols=["DESYNPUF_ID"],
        chunksize=100_000,
        low_memory=False
    ):

        claim_ids = chunk["DESYNPUF_ID"].dropna()

        total_claims += len(claim_ids)

        matched = claim_ids.isin(beneficiary_ids)

        matched_claims += matched.sum()
        unmatched_claims += (~matched).sum()

    match_rate = (
        matched_claims / total_claims * 100
        if total_claims > 0
        else 0
    )

    print(f"\n{name}")
    print(f"Total claims with beneficiary ID : {total_claims:,}")
    print(f"Matched beneficiary IDs          : {matched_claims:,}")
    print(f"Unmatched beneficiary IDs        : {unmatched_claims:,}")
    print(f"Match rate                       : {match_rate:.2f}%")
```

## Beneficiary-to-Claim Relationship Findings

The relationship validation confirms that all claim records can be linked to a beneficiary ID contained in the beneficiary summary data.

### Results

| Dataset | Claims with Beneficiary ID | Matched | Unmatched | Match Rate |
|---|---:|---:|---:|---:|
| Inpatient | 66,773 | 66,773 | 0 | 100.00% |
| Outpatient | 790,790 | 790,790 | 0 | 100.00% |

### Beneficiary Population

The three annual beneficiary files contain a combined set of 116,352 unique `DESYNPUF_ID` values.

The annual files therefore represent repeated annual records for the same underlying beneficiary population rather than three completely separate populations.
